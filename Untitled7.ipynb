```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

print("Libraries imported successfully!")
print("TensorFlow version:", tf.__version__)
```

```python

import os
import zipfile
import urllib.request

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00235/household_power_consumption.zip"
zip_path = "/content/household_power_consumption.zip"
file_path = "/content/household_power_consumption.txt"

if not os.path.exists(file_path):
    print("Downloading dataset...")
    urllib.request.urlretrieve(url, zip_path)

    with zipfile.ZipFile(zip_path, "r") as z:
        z.extractall("/content")

print("Dataset loaded successfully!")
print("File exists:", os.path.exists(file_path))
```

```python

df = pd.read_csv(
    file_path,
    sep=";",
    na_values=["?"],
    low_memory=False
)

df["Global_active_power"] = pd.to_numeric(
    df["Global_active_power"], errors="coerce"
)

df = df.dropna(subset=["Global_active_power"]).reset_index(drop=True)

# Select 10,000 consecutive valid records
data = df["Global_active_power"].iloc[:10000].to_numpy()
data = data.reshape(-1, 1)

print("Selected records:", len(data))
print("Missing values:", np.isnan(data).sum())
print("First five readings:")
print(data[:5])
```

```python

# Split chronologically: 80% training, 20% testing
split = int(len(data) * 0.8)

train_data = data[:split]
scaler = MinMaxScaler()

train_scaled = scaler.fit_transform(train_data)
all_scaled = scaler.transform(data)

# Create sequences using 60 previous readings
window = 60

def create_sequences(values, start, end, window):
    X, y = [], []

    for i in range(start + window, end):
        X.append(values[i-window:i])
        y.append(values[i, 0])

    return np.array(X), np.array(y)

X_train, y_train = create_sequences(
    all_scaled, 0, split, window
)

X_test, y_test = create_sequences(
    all_scaled, split - window, len(data), window
)

print("Training sequences:", X_train.shape)
print("Testing sequences:", X_test.shape)
print("Normalization completed!")
```

```python

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(window, 1)),
    tf.keras.layers.LSTM(32),
    tf.keras.layers.Dense(16, activation="relu"),
    tf.keras.layers.Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mse"
)

history = model.fit(
    X_train, y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.1,
    shuffle=False,
    verbose=1
)

print("LSTM model training completed!")
```

```python

# Predict test readings
pred_scaled = model.predict(X_test, verbose=0)

# Convert values back to kW
y_actual = scaler.inverse_transform(
    y_test.reshape(-1, 1)
).flatten()

y_predicted = scaler.inverse_transform(
    pred_scaled
).flatten()

mae = mean_absolute_error(y_actual, y_predicted)
rmse = np.sqrt(mean_squared_error(y_actual, y_predicted))

print("MAE:", round(mae, 4), "kW")
print("RMSE:", round(rmse, 4), "kW")
```

```python

plt.figure(figsize=(12, 5))

plt.plot(y_actual[:300], label="Actual Consumption")
plt.plot(y_predicted[:300], label="Predicted Consumption")

plt.title("Actual vs Predicted Electricity Consumption")
plt.xlabel("Test Record")
plt.ylabel("Global Active Power (kW)")
plt.legend()
plt.grid(True)
plt.show()
```

```python

results = pd.DataFrame({
    "Actual Consumption (kW)": y_actual[:5],
    "Predicted Consumption (kW)": y_predicted[:5]
})

print("Predictions on unseen test records:")
print(results.round(4).to_string(index=False))
```

```python

last_60 = all_scaled[-window:].reshape(1, window, 1)

next_scaled = model.predict(last_60, verbose=0)
next_power = scaler.inverse_transform(next_scaled)[0, 0]

print("Predicted next-minute consumption:",
      round(float(next_power), 4), "kW")
```
